```python
import pandas as pd
import os

dataset_directory_path = '/root/.cache/kagglehub/datasets/shrutimechlearn/churn-modelling/versions/1'

file_name = 'Churn_Modelling.csv'

full_csv_path = os.path.join(dataset_directory_path, file_name)

df = pd.read_csv(full_csv_path)

df.head()
```

```python
df=df.drop(columns=['RowNumber','CustomerId','Surname'])
```

```python
df
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
le.fit(df['Gender'])
df['Gender']=le.transform(df['Gender'])
```

```python
df
```

```python
from sklearn.preprocessing import OneHotEncoder

OneHot = OneHotEncoder()

df_geo_encoded = OneHot.fit_transform(df[['Geography']])

df_geo_encoded_df = pd.DataFrame(df_geo_encoded.toarray(), columns=OneHot.get_feature_names_out(['Geography']))

df_processed = pd.concat([df.drop('Geography', axis=1), df_geo_encoded_df], axis=1)

display(df_processed.head())
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df_processed.drop('Exited', axis=1), df_processed['Exited'], test_size=0.2, random_state=42)
#
```

```python
print(y_train.value_counts(normalize=True))
```

```python
from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier(
    n_estimators=300,
    class_weight='balanced',
    random_state=42,
    n_jobs=-1
)
rf.fit(X_train, y_train)
```

```python
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score

y_pred = rf.predict(X_test)
y_proba = rf.predict_proba(X_test)[:, 1]

print("Classification report:\n")
print(classification_report(y_test, y_pred))

print("Confusion matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nROC-AUC:", round(roc_auc_score(y_test, y_proba), 3))
```
