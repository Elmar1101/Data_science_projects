```python
df.dropna(subset=['CustomerDOB', 'CustGender', 'CustAccountBalance', 'CustLocation'], inplace=True)

df.info()
```

```python
df.duplicated()
```

```python
df['CustomerDOB'] = pd.to_datetime(df['CustomerDOB'], format='%d/%m/%y', errors='coerce')
df['TransactionDate'] = pd.to_datetime(df['TransactionDate'], format='%d/%m/%y', errors='coerce')

df.info()
```

```python
print("Unique values in CustGender:", df['CustGender'].unique())
print("Number of unique values in CustLocation:", df['CustLocation'].nunique())
# Display top 20 most frequent locations to check for common variations
print("Top 20 most frequent CustLocation values:\n", df['CustLocation'].value_counts().head(20))
```

```python
snapshot_date = df['TransactionDate'].max() + pd.Timedelta(days=1)
print("Snapshot date (referans nogte):", snapshot_date)
```

```python
rfm = df.groupby('CustomerID').agg(
    Recency=('TransactionDate', lambda x: (snapshot_date - x.max()).days),
    Frequency=('TransactionID', 'count'),
    Monetary=('TransactionAmount (INR)', 'sum'),
    AvgTransactionAmount=('TransactionAmount (INR)', 'mean'),
    AvgAccountBalance=('CustAccountBalance', 'mean'),
).reset_index()
```

```python
from sklearn.preprocessing import StandardScaler
rfm = rfm.dropna(subset=['AvgAccountBalance'])
rfm['Monetary_log'] = np.log1p(rfm['Monetary'])
rfm['AvgAccountBalance_log'] = np.log1p(rfm['AvgAccountBalance'])
rfm['AvgTransactionAmount_log'] = np.log1p(rfm['AvgTransactionAmount'])
features = ['Recency', 'Frequency', 'Monetary_log', 'AvgAccountBalance_log']
X = rfm[features].copy()
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
from sklearn.cluster import MiniBatchKMeans
from sklearn.metrics import silhouette_score
rng = np.random.default_rng(42)
sample_idx = rng.choice(X_scaled.shape[0], size=20000, replace=False)
X_sample = X_scaled[sample_idx]

for k in range(2, 8):
    km = MiniBatchKMeans(n_clusters=k, random_state=42, n_init=10, batch_size=1000)
    labels = km.fit_predict(X_scaled)
    sil = silhouette_score(X_sample, km.predict(X_sample))
    print(f"K={k}: inertia={km.inertia_:.0f}, silhouette={sil:.4f}")
```

```python
from sklearn.cluster import MiniBatchKMeans
km = MiniBatchKMeans(n_clusters=4, random_state=42, n_init=10, batch_size=1000)
rfm['Cluster'] = km.fit_predict(X_scaled)

summary = rfm.groupby('Cluster').agg(
    Musteri_sayi=('CustomerID', 'count'),
    Orta_Recency=('Recency', 'mean'),
    Orta_Frequency=('Frequency', 'mean'),
    Orta_Monetary=('Monetary', 'mean'),
    Orta_AvgTransaction=('AvgTransactionAmount', 'mean'),
    Orta_Balans=('AvgAccountBalance', 'mean'),
).round(1)
summary['Faiz'] = (summary['Musteri_sayi'] / summary['Musteri_sayi'].sum() * 100).round(1)

print(summary.to_string())
```
